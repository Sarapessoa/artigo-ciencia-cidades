```python
import pandas as pd 

import numpy as np # biblioteca de analise numerica

import matplotlib.pyplot as plt # biblioteca de plotagem  visual / para graficos
```

```python
dataframe = pd.read_csv("ipeadata[08-09-2025-08-08].csv",sep=";",header=1)
dataframe
```

```python
dataframe["2022"]
```

```python
pop = [x for x in dataframe["2022"]]
plt.loglog()
plt.ylabel("frequencia")
plt.xlabel("populacao")
plt.hist(pop,bins=30)
```

```python
pop = [x for x in dataframe["2022"]]

bins = np.logspace(0,max(np.log10(pop)),30)

hist,bins_edges = np.histogram(pop, bins, density=True)

bins_edges = 0.5*(bins_edges[:-1]+bins_edges[1:])

plt.loglog()
plt.xlim((1e2,1e7))

plt.xlabel("Populacao")
plt.ylabel("Frequencia/Probabilidade")
plt.plot(bins_edges,hist,"o")

#plt.loglog()
```

```python
pop = [x for x in dataframe["2022"]]

bins = np.linspace(1,max(pop),100)

hist,bins_edges = np.histogram(pop, bins, density=True)

bins_edges = 0.5*(bins_edges[:-1]+bins_edges[1:])


plt.xlabel("Populacao")
plt.ylabel("Frequencia/Probabilidade")
plt.plot(bins_edges,hist,"o")

plt.loglog()
```

```python
clean_pop = pop[~np.isnan(pop)]

rank = np.arange(1,len(clean_pop)+1)
y = sorted(clean_pop,reverse=True)

plt.loglog()

plt.plot(rank,y,"o-")
```
